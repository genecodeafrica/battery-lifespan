```python
for cell in battery["cell"].unique():
    cell_data = battery[battery["cell"] == cell].sort_values("cycle")
    
    initial_capacity = cell_data["capacity_mAh"].iloc[0]
    final_capacity = cell_data["capacity_mAh"].iloc[-1]
    
    capacity_loss = initial_capacity - final_capacity
    percentage_loss = (capacity_loss / initial_capacity) * 100
    
    print(f"{cell}:")
    print(f"  Initial capacity: {initial_capacity:.2f} mAh")
    print(f"  Final capacity:   {final_capacity:.2f} mAh")
    print(f"  Capacity loss:     {capacity_loss:.2f} mAh")
    print(f"  Percentage loss:   {percentage_loss:.2f}%")
    print()
```

```python
eol_threshold = 0.80

eol_results = []

for cell in battery["cell"].unique():
    cell_data = battery[battery["cell"] == cell].sort_values("cycle").copy()
    
    initial_capacity = cell_data["capacity_mAh"].iloc[0]
    threshold_capacity = initial_capacity * eol_threshold
    
    eol_data = cell_data[
        cell_data["capacity_mAh"] <= threshold_capacity
    ]
    
    if len(eol_data) > 0:
        eol_cycle = eol_data["cycle"].iloc[0]
    else:
        eol_cycle = None
    
    eol_results.append({
        "cell": cell,
        "initial_capacity_mAh": initial_capacity,
        "eol_capacity_mAh": threshold_capacity,
        "eol_cycle": eol_cycle,
        "last_observed_cycle": cell_data["cycle"].iloc[-1]
    })

eol_df = pd.DataFrame(eol_results)

eol_df
```

```python
battery_rul = battery.copy()

eol_map = eol_df.set_index("cell")["eol_cycle"]

battery_rul["eol_cycle"] = battery_rul["cell"].map(eol_map)

battery_rul["rul_cycles"] = (
    battery_rul["eol_cycle"] - battery_rul["cycle"]
)

battery_rul.head(10)
```

```python

battery_ml = battery_rul.copy()


battery_ml = battery_ml.sort_values(
    ["cell", "cycle"]
).reset_index(drop=True)



initial_capacity_map = (
    battery_ml
    .groupby("cell")["capacity_mAh"]
    .first()
)

battery_ml["initial_capacity_mAh"] = (
    battery_ml["cell"].map(initial_capacity_map)
)




battery_ml["capacity_retention"] = (
    battery_ml["capacity_mAh"] /
    battery_ml["initial_capacity_mAh"]
)



battery_ml["capacity_loss_mAh"] = (
    battery_ml["initial_capacity_mAh"] -
    battery_ml["capacity_mAh"]
)




battery_ml["capacity_change_mAh"] = (
    battery_ml
    .groupby("cell")["capacity_mAh"]
    .diff()
)

battery_ml["cycle_change"] = (
    battery_ml
    .groupby("cell")["cycle"]
    .diff()
)

battery_ml["degradation_rate_mAh_per_cycle"] = (
    battery_ml["capacity_change_mAh"] /
    battery_ml["cycle_change"]
)


battery_ml["degradation_rate_mAh_per_cycle"] = (
    battery_ml["degradation_rate_mAh_per_cycle"]
    .fillna(0)
)




features = [
    "cycle",
    "capacity_mAh",
    "duration_s",
    "start_voltage_V",
    "end_voltage_V",
    "avg_temp_C",
    "n_samples",
    "initial_capacity_mAh",
    "capacity_retention",
    "capacity_loss_mAh",
    "degradation_rate_mAh_per_cycle"
]




X = battery_ml[features].copy()

y = battery_ml["rul_cycles"].copy()



print("FEATURES")
print("------------------------------")

for i, feature in enumerate(features, 1):
    print(f"{i}. {feature}")

print("\nNumber of observations:", len(X))
print("Number of features:", len(features))

print("\nX shape:", X.shape)
print("y shape:", y.shape)

print("\nMissing values in X:")
print(X.isnull().sum())

print("\nMissing values in y:")
print(y.isnull().sum())

print("\nRUL statistics:")
print(y.describe())
```

```python
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

print("Machine learning libraries loaded successfully.")
```

```python
features = [
    "cycle",
    "capacity_mAh",
    "duration_s",
    "start_voltage_V",
    "end_voltage_V",
    "avg_temp_C",
    "n_samples",
    "initial_capacity_mAh",
    "capacity_retention",
    "capacity_loss_mAh",
    "degradation_rate_mAh_per_cycle"
]

X = battery_ml[features].copy()
y = battery_ml["rul_cycles"].copy()


groups = battery_ml["cell"].copy()

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("Number of battery cells:", groups.nunique())
print("\nFeatures:")
for feature in features:
    print("-", feature)
```

```python
gkf = GroupKFold(n_splits=8)

print("GroupKFold created successfully.")
print("Number of folds:", gkf.get_n_splits())
```

```python
rf_predictions = np.zeros(len(y))

for train_idx, test_idx in gkf.split(X, y, groups):

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]

    rf_model = RandomForestRegressor(
        n_estimators=300,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    )

    rf_model.fit(X_train, y_train)

    rf_predictions[test_idx] = rf_model.predict(X_test)

print("Random Forest training completed.")
```

```python
rf_mae = mean_absolute_error(y, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y, rf_predictions))
rf_r2 = r2_score(y, rf_predictions)

print("Random Forest Results")
print("---------------------")
print(f"MAE  : {rf_mae:.2f} cycles")
print(f"RMSE : {rf_rmse:.2f} cycles")
print(f"R²   : {rf_r2:.4f}")
```

```python
gb_predictions = np.zeros(len(y))

for train_idx, test_idx in gkf.split(X, y, groups):

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]

    gb_model = GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=2,
        random_state=42
    )

    gb_model.fit(X_train, y_train)

    gb_predictions[test_idx] = gb_model.predict(X_test)

print("Gradient Boosting training completed.")
```

```python
gb_mae = mean_absolute_error(y, gb_predictions)
gb_rmse = np.sqrt(mean_squared_error(y, gb_predictions))
gb_r2 = r2_score(y, gb_predictions)

print("Gradient Boosting Results")
print("-------------------------")
print(f"MAE  : {gb_mae:.2f} cycles")
print(f"RMSE : {gb_rmse:.2f} cycles")
print(f"R²   : {gb_r2:.4f}")
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        rf_mae,
        gb_mae
    ],
    "RMSE": [
        rf_rmse,
        gb_rmse
    ],
    "R2": [
        rf_r2,
        gb_r2
    ]
})

comparison
```

```python


from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd
import numpy as np
import joblib
import os


features = [
    "cycle",
    "capacity_mAh",
    "duration_s",
    "start_voltage_V",
    "end_voltage_V",
    "avg_temp_C",
    "n_samples",
    "initial_capacity_mAh",
    "capacity_retention",
    "capacity_loss_mAh",
    "degradation_rate_mAh_per_cycle"
]

X = battery_ml[features].copy()
y = battery_ml["rul_cycles"].copy()

print("Number of observations:", len(X))
print("Number of features:", len(features))
print("\nFeatures:")
for feature in features:
    print("-", feature)
```

```python


final_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=2,
    random_state=42
)


final_model.fit(X, y)

print("Final Gradient Boosting model trained successfully.")
```

```python


final_predictions = final_model.predict(X)

final_mae = mean_absolute_error(y, final_predictions)
final_rmse = np.sqrt(mean_squared_error(y, final_predictions))
final_r2 = r2_score(y, final_predictions)

print("FINAL MODEL TRAINING PERFORMANCE")
print("----------------------------------")
print(f"MAE  : {final_mae:.2f} cycles")
print(f"RMSE : {final_rmse:.2f} cycles")
print(f"R²   : {final_r2:.4f}")
```

```python
model_directory = "../ml/models"

os.makedirs(model_directory, exist_ok=True)

print("Model directory ready:", model_directory)
```

```python
model_path = os.path.join(
    model_directory,
    "battery_lifespan_model.joblib"
)

joblib.dump(final_model, model_path)

print("Final model saved successfully.")
print("Location:", model_path)
```

```python
# --------------------------------------------
# 6. Save model package
# --------------------------------------------

model_package = {
    "model": final_model,

    "features": features,

    "target": "rul_cycles",

    "eol_threshold": 0.80,

    "model_name": "Gradient Boosting Regressor",

    "parameters": {
        "n_estimators": 200,
        "learning_rate": 0.05,
        "max_depth": 2,
        "random_state": 42
    },

    "validation_metrics": {
        "MAE": 669.736465,
        "RMSE": 875.951112,
        "R2": 0.857924
    }
}

package_path = os.path.join(
    model_directory,
    "battery_lifespan_model_package.joblib"
)

joblib.dump(model_package, package_path)

print("Model package saved successfully.")
print("Location:", package_path)
```
